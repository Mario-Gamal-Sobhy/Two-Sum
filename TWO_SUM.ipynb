```python
class Solution(object):
    def twoSum(self, nums, target):
        hash_map = {}
        for index, number in enumerate(nums):
            complement = target - number
            if complement in hash_map:
                return [hash_map[complement], index]
            hash_map[number] = index
        return [] # Return an empty list if no two sum solution is found
```

```python
solution = Solution()

# Test Case 1: Basic example
nums1 = [2, 7, 11, 15]
target1 = 9
result1 = solution.twoSum(nums1, target1)
print(f"Test Case 1: nums={nums1}, target={target1}, result={result1} (Expected: [0, 1])")

# Test Case 2: Numbers with different order
nums2 = [3, 2, 4]
target2 = 6
result2 = solution.twoSum(nums2, target2)
print(f"Test Case 2: nums={nums2}, target={target2}, result={result2} (Expected: [1, 2])")

# Test Case 3: No solution
nums3 = [3, 2, 4]
target3 = 10
result3 = solution.twoSum(nums3, target3)
print(f"Test Case 3: nums={nums3}, target={target3}, result={result3} (Expected: [])")

# Test Case 4: Duplicate numbers, different indices
nums4 = [3, 3]
target4 = 6
result4 = solution.twoSum(nums4, target4)
print(f"Test Case 4: nums={nums4}, target={target4}, result={result4} (Expected: [0, 1])")
```
